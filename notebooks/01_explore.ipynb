# 01 — Explore a PDF
Render a few pages from `data/raw/` and print the native text layer. No LLM calls.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))

import matplotlib.pyplot as plt
from PIL import Image
from src.config import get_settings
from src.ingest import find_pdfs, ingest_pdf

settings = get_settings()
pdfs = find_pdfs(settings.raw_dir)
print(f'{len(pdfs)} PDF(s) in {settings.raw_dir}')
for p in pdfs[:20]:
    print(' ', p.name)
```

```python
PDF_INDEX = 0   # which PDF to look at
N_PAGES = 3     # how many pages to render
DPI = 100       # low DPI is fine for browsing

pages = ingest_pdf(pdfs[PDF_INDEX], dpi=DPI, limit=N_PAGES)
print(pdfs[PDF_INDEX].name, '->', len(pages), 'pages rendered')
```

```python
for p in pages:
    fig, ax = plt.subplots(figsize=(10, 13))
    ax.imshow(Image.open(p.image_path), cmap='gray')
    ax.set_title(f'page {p.page_number}  ({p.width}x{p.height}px)')
    ax.axis('off')
    plt.show()
```

```python
for p in pages:
    print(f'===== page {p.page_number}: {len(p.text)} chars of native text =====')
    print(p.text[:2000] or '(no text layer: scanned image)')
    print()
```
